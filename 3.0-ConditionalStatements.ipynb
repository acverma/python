```python
# else
# the else statement is optional
# if the condition is not met, the else block will be executed

age = 16
if age >= 18:
    print("You are an adult.")
else:
    print("You are a minor.")
```

```python
## elif statement
# The elif statement allows you to check multiple conditions in a single if-else block.

age = int(input("Enter your age: "))
if age < 13:
    print("You are a child.")
elif age < 18:
    print("You are a teenager.")
elif age < 65:
    print("You are an adult.")
else:
    print("You are a senior citizen.")

```

```python
## Nested Conditional Statements
## You can nest if statements within each other to create more complex conditions.

age = int(input("Enter your age: "))
if age >= 18:
    print("You are an adult.")
    if age < 65:
        print("You are a working-age adult.")
    else:
        print("You are a senior citizen.")
else:
    print("You are a minor.")
```

```python
num = int(input("Enter the number: "))
if num >= 0:
    print("The number is positive.")
    if num %2 == 0:
        print("The number is even.")
    else:
        print("The number is odd.")
else:
    print("The number is negative.")
```

```python
# Use cases Practical examples

# 1. Grading System
score = float(input("Enter your score: "))
if score >= 95:
    print("You got an A++.")
elif score >= 90 & score < 95:
    print("You got an A+.")
elif score >= 90:
    print("You got an A.")
elif score >= 80 & score < 90:
    print("You got a B.")
elif score >= 70:
    print("You got a C.")
elif score >= 60:
    print("You got a D.")
else:
    print("You failed.")
```

```python
## Determine if a year is a leap year using nested if statements
year = int(input("Enter a year: "))
if (year%4 == 0):
    if (year%100 == 0):
        # If the year is divisible by 100, it must also be divisible by 400 to be a leap year

        if (year%400 == 0):
            print(f"{year} is a leap year.")
        else:
            print(f"{year} is not a leap year.")
    else:
        print(f"{year} is a leap year.")
else:
    print(f"{year} is not a leap year.")
```

```python
n = 10
sum = 0
for i in range(1, n + 1):
    sum += i

print(f"The sum of the first {n} natural numbers is: {sum}")
```

```python
## Example: Check if a number is prime
num = int(input("Enter a number: "))
is_prime = True
if num < 2:
    is_prime = False
for i in range(2, int(num**0.5) + 1):
    if num % i == 0:
        is_prime = False
        break
if is_prime:
    print(f"{num} is a prime number.")
else:
    print(f"{num} is not a prime number.")
```
