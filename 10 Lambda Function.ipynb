## Lambda Functions in Python

### Lambda functions are anonymous functions defined using *lmabda* keyword.
- It can have any number of arguments but only one expression
- They are commonly used for short operations or as an arguments to higher-order functions.
  

```python
## Syntax

# lambda arguments: expression #ignore
```

```python
def addition(a,b):
    return a + b
```

```python
addition(2,3)
```

```python
addition = lambda a,b : a + b
type(addition)
print(addition(5,6))
```

```python
def even(num):
    if num%2==0:
        return True
```

```python
even(24)
```

```python
even1 = lambda num:num%2==0
even1(12)
```

```python
def addition(x,y,z):
    return x + y + z

addition(12,13,14)
```

```python
add = lambda x,y,z : x + y + z
add(12,15,16)
```

### MAP

```python
## map() -> applies a function to all items in a list
numbers = [1,2,3]
def sq(num):
    return num**2
sq(4)
```

```python
## can we have a better way to do the square 

## YES with map
numbers = [1,2,3]
list(map(lambda x:x**2,numbers))
```
