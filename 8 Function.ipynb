```python
def even_odd(num):
    """ this function find even or odd
    Args:
        num (int): number to be checked
    Returns:
        str: "even" if num is even else "odd"
    """
    if num%2==0:
        print(f" the number {num} is even")
    else:
        print(f" the number {num} is odd")


```

```python
# Call the function with an argument
even_odd(24) 

even_odd(23)
```

```python
## Default parameters
def greet(name):
    """ this function greets a person with given name
    Args:
        name (str): name of the person
    Returns:
        str: greeting message
    """
    print(f" Hello, {name}! Welcome aboard.")
```

```python
greet("Abhishek")
greet()
```

```python
def greetings(name="Guest"):
    print(f" Hello, {name}! Welcome aboard.")
greetings("Abhishek")
greetings()
```

```python
## Postional and Keyword Arguments
# Positional arguments are passed to function in correct positional order.
def print_numbers(*args):
    """ this function prints numbers passed as arguments
    Args:
        *args: variable length argument list
    Returns:
        None
    """
    for num in args:
        print(num, end=" ")
    # print()
print_numbers(1,2,3,4,5)
```

```python
# Keyword arguments are passed to function by explicitly specifying parameter name.

def print_info(**kwargs):
    """ this function prints key-value pairs passed as keyword arguments
    Args:
        **kwargs: variable length keyword argument list
    Returns:
        None
    """
    for key, value in kwargs.items():
        print(f"{key}: {value}")
print_info(name="Abhishek", age=25, city="New York")

```

```python
def student_info(name, age, grade):
    """ this function prints student information
    Args:
        name (str): name of the student
        age (int): age of the student
        grade (str): grade of the student
    Returns:
        None
    """
    print(f"Name: {name}, Age: {age}, Grade: {grade}")

student_info(age=20, name="Abhishek", grade="A")
```

```python
## combine both positional and keyword arguments
def print_details(*args,**kwargs):
    """ this function prints details using both positional and keyword arguments
    Args:
        *args: variable length argument list
        **kwargs: variable length keyword argument list
    Returns:
        None
        """
   

    for val in args:
        print(f"positional arguments : {val}", end = " ")
    for key in kwargs:
        print(f"keyword arguments : {key}", end = "\n")

    for key,value in kwargs.items():
        print(f"{key} : {value}")

print_details(1,2,3,name="Abhishek", age=25)
```

```python
## Return Statement
def add(a,b):
    """ this function adds two numbers
    Args:
        a (int): first number
        b (int): second number
    Returns:
        int: sum of a and b
    """
    sum = a + b
    return sum
```

```python
add(3,5)
```
