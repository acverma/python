### Dictionary 

#### These are unordered collection of Items.
#### They store data in Key-value pairs.
#### Key must be uniques and immutable(strings, numbers or tuples)

```python
## Creating Dictionaries

empty_dict = {}
print(type(empty_dict))
```

```python
empty_dict = dict()
print(type(empty_dict))
print(empty_dict)
```

```python
student = {
    "name": "John",
    "age": 20,
    "courses": ["Math", "Science"]
}
print(student)
```

```python
# #error
# student = {
#     "name": "John",
#     "age": 20,}
# print(student)
```

```python
## accessing dictionary items

print(student["name"])
print(student["age"])
print(student["courses"])
```

```python
# accessing Dictionary elements

student = {
    "name": "John",
    "age": 20,
    "courses": ["Math", "Science"],
    "grade" : "A"
}

print(student['age'])
print(student["grade"])
```

```python
## another way using defaul get() method
print(student.get("grade"))
print(student.get("address", "Not Found"))  # returns default value if key not found
```

```python
# Modifying Dictionary items
## Dictionary are mutable, s we can change the value of an existing key
print(student)
```

```python
##
student["age"] = 21 # updated the existing key
print(student["age"])
```

```python
student["address"] = "USA" ## added a new key-value pair
print(student)
```

```python
del student['grade']
print(student)
```

```python
## common dictionary methods

keys = student.keys() # get all the keys
print(keys)
```

```python
values = student.values() # get all the values
print(values)
```

```python
items = student.items() # get all the key-value pairs
print(items)
```

```python
# Shallow copy a dictionary
new_student = student.copy()
print(new_student)
```

```python
student['name'] = "Jane"
print(student)
print(new_student)  # new_student remains unchanged
```

```python
# iterating through a dictionary
for key in student.keys():
    print(f"{key}: {student[key]}")
```

```python
for value in student.values():
    print(value)
```

```python
for key, value in student.items():
    print(f"{key}: {value}")
```

```python
# Nested Dictionaries
students = {
    "student1": {"student_id" : 1,"name" : "Abhishek", "age" : 21, "courses" : ["Math", "Science"]},
    "student2" : {"student_id" : 2,"name" : "PeterParker", "age" : 22, "courses" : ["English", "History"]}

}
print(students)
```

```python
# Access nested dictionary items
print(students["student1"]["name"])
print(students["student1"]["courses"])
print("\n")
print(students["student2"]["name"])
print(students["student2"]["courses"])
```

```python
student.items()
```

```python
# Iterating over nested dictionaries
for student_id, student_info in students.items():
    print(f"(sudent_id): {student_info}")
    # Accessing nested dictionary items
    for key,value in student_info.items():
        print(f"{key}: {value}")
```

```python
# Nested Dictionaries
students = {
    "student1": {"student_id" : 1,"name" : "Abhishek", "age" : 21, "courses" : ["Math", "Science"]},
    "student2" : {"student_id" : 2,"name" : "PeterParker", "age" : 22, "courses" : ["English", "History"]}

}
```

```python
list_students = list(map(lambda x : x['name'], students.values()))
print(list_students)
```

```python
def students_details(students):
    for key,value in students.items():
        # return f"({key} : {value})"
        print(f"({key} : {value})")
```

```python
list_students = list(map(students_details ,students.values()))
print(list_students)
```

```python
filter_student = list(filter(lambda x : x['age'] > 21, students.values()))
print(filter_student)
```

```python
numbers = [1,1,2,2,2,3,3,3,4,4,4,4,4,4]
for i in numbers:
    for j in numbers:
        if i != j:
            if i + j == 4:
                print(f"({i}, {j})")
```

```python
numbers = [1,2,2,2,3,3,3,4,4,4,4,4,4]
frequency = {}
for number in numbers:
    if number in frequency:
        frequency[number] += 1
    else:
        frequency[number] = 1
print(frequency)
```

```python
### Merge two dictionaries into one
dict1 = {'a': 1, 'b' : 2}
dict2 = {'b': 3, 'c' : 4}
merg_dict = {**dict1, **dict2}
print(merg_dict)
```
