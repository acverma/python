## List Comprehension

### Most important data structure for memory

```python
list = []
for x in range(10):
    list.append(x*2)
print(list)
    
```

```python
## we can write above code in a single line using list comprehension

list = [x*2 for x in range(10)]
print(list)
```

```python
[x*2 for x in range(10) if x % 2 == 0]
```

```python
[x**2 for x in range(10)]
```

```python
[x**2 for x in range(10) if x % 2 ==0]
```

### Basic Syntax -> [expression for item in iterable]

###  with conditional logic -> [expression for item in iterable if condition]
### Nested List comprehension -> [expression for item1 in iterable1 for item2 in iterable2]

```python
list = []
for i in range(10):
    if i % 2 == 0:
        list.append(i)
print(list)
```

```python
[i for i in range(10) if i % 2 ==0]
```

```python
[num for num in range(10) if num % 2 == 0]
```

```python
## iterable nested list comprehension

list1 = [1,2,3,4]
list2 = ['a','b','c','d']

pair = [[i,j] for i in list1 for j in list2]
print(pair)
```

```python
## List Comprehension with function calls

words = ['hello', 'world', 'python', 'rocks']
length = [len(word) for word in words]
print(length)
```

```python
## Real world Examples

# Create a To Do List to Keep Track of Tasks

to_do_list = ["Buy groceries", "Clean the house", "Finish homework", "Call mom"]

# Add a new task
to_do_list.append("Schedule a meeting")
to_do_list.append("Prepare presentation")


# Remove a completed task
to_do_list.remove("Clean the house")

# Check if a task is in the list
if "Call mom" in to_do_list:
    print("You need to call mom!")

print("Current To Do List:")
for task in to_do_list:
    print(f"- {task}")
```

```python
# Oraganizing a list to store and calculate average grades for students

grades = [85, 90, 78, 92, 88]

# Adding a new grade
grades.append(95)

# Calculating the average grade
average_grade = sum(grades) / len(grades)
print(f"Average grade: {average_grade}")

# Finding the highest and lowest grades
highest_grade = max(grades)
lowest_grade = min(grades)
print(f"Highest grade: {highest_grade}")
print(f"Lowest grade: {lowest_grade}")
```

```python
# managing an Inventory List for a Store
inventory = ["Apples", "Bananas", "Oranges", "Grapes"]

# Adding a new item to the inventory
inventory.append("Mangoes")

# Removing an item from the inventory
inventory.remove("Bananas")

# Displaying the current inventory
print("Current Inventory:", inventory)
```

```python
# checking if an item is in the inventory
item = "Oranges"
if item in inventory:
    print(f"{item} is available in the inventory.")
else:
    print(f"{item} is not available in the inventory.")
# Displaying the current inventory
print("Current Inventory:")
for item in inventory:
    print(f"- {item}")
    
```

# Counting speicific items in a list




```python
positive_feedback_count = sum(1 for comment  in feedback if "good" in comment.lower() or "excellent" in comment.lower())

print(f"Number of positive feedback comments: {positive_feedback_count}")

# Print all the feedback comments
print(" User Feedback Comments:")
for comment in feedback:
    print(f"- {comment}")
    
```

```python
from numpy import negative


feedback = ['Great Service !', 'Good Quality', 'Average Experience', 'Excellent Support', 'Poor Packaging','Very Satisfied', 'Could be better','Excellent Service']

# Adding new feedback
feedback.append('Not happy with the Service')
feedback.append('Fast Delivery')  
                

# Counting specific feedback
pos_keywords = ["good", "excellent", "great", "satisfied"]
neg_keywords = ["poor", "not happy", "could be better"]

positive_feedback_count = sum(1 for comment in feedback if any (pos_keywords in comment.lower() for pos_keywords in pos_keywords))
negative_feedback_count = sum(1 for comment in feedback if any (neg_keywords in comment.lower() for neg_keywords in neg_keywords))
print(f"Number of negative feedback comments: {negative_feedback_count}")

# positive_feedback_count = sum(1 for comment in feedback if "great" in comment.lower() or "excellent" in comment.lower() or "good")
print(f"Number of postive feedback comments: {positive_feedback_count}")

# Print all feedback comments
print("All Feedback Comments:")
for comment in feedback:
    print(f" - {comment}", end = '')

```

```python
employee = ['Mr Bean', 'Mr Trump', 'Mr Modi', 'Mr Musk', 'Mr Gates']
department = ['HR', 'Finance', 'IT', 'R&D', 'Sales']
salary = [50000, 120000, 90000, 150000, 200000]

# Create a list of employee, department, and salary using list comprehension
employee_map = [[employee[i], department[i], salary[i]] for i in range (len(employee))]
print(employee_map)
```

```python
emp_map = list(zip(employee, department, salary))
print(emp_map)
```
