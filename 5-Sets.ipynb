```python
# Mathematical operations on sets
set1 = {1, 2, 3, 4, 5, 6}
set2 = {4, 5, 6, 7, 8, 9}
print("Union ->",set1.union(set2))  # Union
print("Intersection ->",set1.intersection(set2))  # Intersection
print("Difference ->",set1.difference(set2))  # Difference
print("Symmetric Difference ->",set1.symmetric_difference(set2))  # Symmetric Difference
my_set = ()  # IGNORE
# Set operations
print("Subset ->",set1.issubset(set2))  # Subset
print("Superset ->",set1.issuperset(set2))  # Superset
print("Disjoint Sets ->",set1.isdisjoint(set2))  # Disjoint sets
```

```python
### Counting unique words in a text

text = " In this tuorial we are discussing about sets. Sets are a collection of unique elements. In this tutorial we will learn about sets in Python. Sets are unordered collections of unique elements. In this tutorial we will learn about sets in Python. Sets are unordered collections of unique elements."

words = text.split()
unique_words = set(words)

print("Unique words in the text:", unique_words)
print("Number of unique words:", len(unique_words))


```
